# QR Code Detection

```python
import cv2 as cv
import numpy as np
```

```python
qrcode_img = cv.imread("images/qrcode1.png")
```

```python
detector = cv.QRCodeDetector()
value, box, _ = detector.detectAndDecode(qrcode_img)

print(value)

box = box.astype(int)
pt1 = (box[0][0][0], box[0][0][1])
pt2 = (box[0][2][0], box[0][2][1])
```

```python
cv.rectangle(qrcode_img, pt1, pt2, (0, 0, 255), 2)
cv.imshow("qrcode", qrcode_img)
cv.waitKey()
cv.destroyAllWindows()
```

# Barcode Detection

```python
import pyzbar.pyzbar as pyzbar
from matplotlib import pyplot as plt
```

```python
barcode_img = cv.imread("images/barcode1.png")
decoded = pyzbar.decode(barcode_img)
```

```python
def display(img, decoded_objs):
    for decoded_obj in decoded_objs:
        points = decoded_obj.polygon

        if len(points) > 4:
            hull = cv.convexHull(np.array([point for point in points], dtype=np.float32))
            hull = list(map(tuple, np.squeeze(hull)))
        else:
            hull = points

        n = len(hull)
        for j in range(0, n):
            cv.line(img, hull[j], hull[(j + 1) % n], (0, 0, 255), 3)
    return img
```

```python
def decode(img):
    decoded_objects = pyzbar.decode(img)

    for obj in decoded_objects:
        print("Type: ", obj.type)
        print("Data: ", obj.data, "\n")
    return decoded_objects
```

```python
print(decode(barcode_img))

cv.imshow("barcode Image", display(barcode_img, decoded))
cv.waitKey()
cv.destroyAllWindows()
```

```python
cap = cv.VideoCapture(0)

while True:
    rec, frame = cap.read()
    frame = cv.resize(frame, (400, 680))
    # objects_decoded = decode(frame)
    objects_decoded = pyzbar.decode(frame)
    display(frame, objects_decoded)

    cv.imshow("frame", frame)

    key = cv.waitKey(1) & 0xFF
    if key == 27:
        cap.release()
        cv.destroyAllWindows()
        break
```
