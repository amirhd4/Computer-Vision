```python
import cv2 as cv
import numpy as np
from cvzone.FaceDetectionModule import FaceDetector
```

```python
cap = cv.VideoCapture(0)
detector = FaceDetector()
```

```python
while True:
    ret, frame = cap.read()
    frame = cv.resize(frame, (480, 640))

    frame, box = detector.findFaces(frame)
    if box:
        center = box[0]["center"]
        cv.circle(frame, center, 5, (255, 8, 255), cv.FILLED)
    cv.imshow("frame", frame)
    key = cv.waitKey(1) & 0xFF
    if key == 27:
        cv.destroyAllWindows()
        cap.release()
        break
```

## Face Mesh Detector

```python
from cvzone.FaceMeshModule import FaceMeshDetector
```

```python
cap = cv.VideoCapture(0)
detector = FaceMeshDetector(maxFaces=1)
```

```python

while True:
    ret, frame = cap.read()
    frame = cv.resize(frame, (480, 640))

    frame, faces = detector.findFaceMesh(frame)

    cv.imshow("frame", frame)
    key = cv.waitKey(1) & 0xFF
    if key == 27:
        cv.destroyAllWindows()
        cap.release()
        break
```
