```python
from imageai.Detection import ObjectDetection
import os
```

```python
detector = ObjectDetection()
detector.setModelTypeAsYOLOv3()
```

```python
model_path = "./deep-learning/models/"
images_path = "./opencv-notebook/images/"
```

```python
detector.setModelPath(os.path.join(model_path, "yolov3.pt"))
detector.loadModel()
```

```python
detection = detector.detectObjectsFromImage(input_image=os.path.join(images_path, "05_tehran.jpg"), output_image_path=os.path.join(images_path, "05_tehran_detected.jpg"))
```
