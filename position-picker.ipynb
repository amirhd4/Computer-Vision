```python
import cv2 as cv
import pickle
```

```python
width, height = 107, 48
```

```python
try:
    with open("CarParkPos", "rb") as f:
        posList = pickle.load(f)
except:
    posList = []
```

```python
def mouse_click(event, x, y, flags, params):
    if event == cv.EVENT_LBUTTONDOWN:
        posList.append((x, y))
    if event == cv.EVENT_RBUTTONDOWN:
        for i, pos_i in enumerate(posList):
            x1, y1 = pos_i
            if x1 < x < x1 + width and y1 < y < y1 + height:
                posList.pop(i)

    with open("CarParkPos", "wb") as file:
        pickle.dump(posList, file)
```

```python
cv.namedWindow("CarPark")
cv.setMouseCallback("CarPark", mouse_click)

while True:
    img = cv.imread("./notebook/images/carPark.png")
    for pos in posList:
        cv.rectangle(img, pos, (pos[0] + width, pos[1] + height), (255, 0, 0), 2)

    cv.imshow("CarPark", img)
    if cv.waitKey(20) & 0xFF == 27:
        break

cv.destroyAllWindows()
```
