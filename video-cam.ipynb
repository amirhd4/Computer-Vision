```python
import cv2 as cv
import numpy as np
```

```python
cap = cv.VideoCapture(0)
```

```python
while True:
    rec, frame = cap.read()
    resized_frame = cv.resize(frame, (400, 600))
    frame_hsv = cv.cvtColor(resized_frame, cv.COLOR_BGR2HSV)

    lower_color = np.array([100, 50, 50])
    upper_color = np.array([116, 255, 180])

    mask_color = cv.inRange(frame_hsv, lower_color, upper_color)
    frame_masked = cv.bitwise_and(resized_frame, resized_frame, mask=mask_color)

    cv.imshow('frame', resized_frame)
    cv.imshow("mask_red", mask_color)
    cv.imshow("frame_maksed", frame_masked)
    key = cv.waitKey(1) & 0xFF
    if key == 27:
        cv.destroyAllWindows()
        cap.release()
        break
```
