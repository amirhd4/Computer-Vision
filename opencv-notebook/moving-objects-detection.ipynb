```python
import numpy as np
import cv2 as cv
```

```python
cap = cv.VideoCapture("./opencv-notebook/Videos/street_camera.mp4")
fps = cap.get(cv.CAP_PROP_FPS)
print(fps)
```

```python
while True and cap.isOpened():
    ret, frame = cap.read()
    ret2, frame2 = cap.read()
    if not (ret and ret2):
        break
    frame = cv.resize(frame, (680, 400))
    frame2 = cv.resize(frame2, (680, 400))

    frame_diff = cv.cvtColor(cv.absdiff(frame, frame2), cv.COLOR_BGR2GRAY)
    blurred_frame = cv.GaussianBlur(frame_diff, (9, 9), 1)
    _, mask = cv.threshold(blurred_frame, 10, 255, cv.THRESH_BINARY)

    contours, _ = cv.findContours(mask, cv.RETR_EXTERNAL, cv.CHAIN_APPROX_SIMPLE)

    for cnt in contours:
        if cv.contourArea(cnt) > 1000:
            (x, y, w, h) = cv.boundingRect(cnt)
            cv.rectangle(frame, (x + w - 100, y + h - 50), (x + w, y + h), (0, 0, 255), 2)
    cv.imshow('frame', frame)
    cv.imshow('frame_diff', frame_diff)
    cv.imshow('frame_mask', mask)

    key = cv.waitKey(10) & 0xFF
    if key == 27:
        break

cv.destroyAllWindows()
cap.release()
```
