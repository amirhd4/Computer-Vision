# Template Matching

```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
image = cv.imread("images/building1.jpg")
template = cv.imread("images/building1_temp.jpg")
w, h, _ = template.shape

result = cv.matchTemplate(image, template, cv.TM_CCOEFF_NORMED)
```

```python
threshold = 0.9

locations = np.where(result >= threshold)

for point in zip(*locations[::-1]):
    cv.rectangle(image, point, (point[0] + h, point[1] + w), (255, 255, 0), 2)
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(121), plt.imshow(image[..., ::-1]), plt.title("planes")
plt.subplot(122), plt.imshow(result[..., ::-1]), plt.title("planes matched")
plt.show()
```
